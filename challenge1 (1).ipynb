```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

simulator = AerSimulator()
SHOTS = 100
```

```python
# Circuit A: One X gate
circuit_a = QuantumCircuit(1, 1)
circuit_a.x(0)
circuit_a.measure(0, 0)

# Circuit B: Two X gates
circuit_b = QuantumCircuit(1, 1)
circuit_b.x(0)
circuit_b.x(0)
circuit_b.measure(0, 0)

# Circuit C: One H gate
circuit_c = QuantumCircuit(1, 1)
circuit_c.h(0)
circuit_c.measure(0, 0)

print("Circuit A:")
print(circuit_a.draw())

print("Circuit B:")
print(circuit_b.draw())

print("Circuit C:")
print(circuit_c.draw())
```

```python
predictions = {
    "Circuit A": "Outcome 1 with 100% probability",
    "Circuit B": "Outcome 0 with 100% probability",
    "Circuit C": "Outcome 0 or 1, each with 50% probability"
}

for circuit, prediction in predictions.items():
    print(f"{circuit}: {prediction}")
```

```python
circuits = {
    "Circuit A": circuit_a,
    "Circuit B": circuit_b,
    "Circuit C": circuit_c
}

results = {}

for name, circuit in circuits.items():
    compiled_circuit = transpile(circuit, simulator)
    job = simulator.run(compiled_circuit, shots=SHOTS)
    counts = job.result().get_counts()
    results[name] = counts

    print(f"\n{name}")
    print("Measurement results:", counts)
    print("Total shots:", sum(counts.values()))
```

```python
for name, counts in results.items():
    plot_histogram(
        counts,
        title=f"{name} - Measurement Results (100 Shots)"
    )
    plt.show()
```

```python
for name, counts in results.items():
    total = sum(counts.values())
    zero_count = counts.get("0", 0)
    one_count = counts.get("1", 0)

    print(f"\n{name}")
    print(f"Prediction: {predictions[name]}")
    print(f"Measured 0: {zero_count}/{total}")
    print(f"Measured 1: {one_count}/{total}")

    if name == "Circuit A":
        print("Comparison: Expected 1; the simulator should return 1 in every shot.")
    elif name == "Circuit B":
        print("Comparison: Expected 0; the simulator should return 0 in every shot.")
    else:
        print("Comparison: Both outcomes are possible; the counts should be approximately balanced.")
```
